```python
# file handling

try:
    balance = 5000
    amount = int(input("Enter Withdraw amount:"))

    if amount > balance:
        raise ValueError("Insufficient balance")

    print("Withdrawal successful")
    print("Remaining balance:", balance - amount)

except ValueError as e:
    print("Error:", e)
```
