```python

def cart_total(cart,prices):
    total=0

    for item in cart:
        if item in prices:
            total=total+prices[item]
    return total

cart = (["apple", "apple", "pear"])
prices = {"apple": 2, "pear": 3}

print(cart_total(cart,prices))
```
