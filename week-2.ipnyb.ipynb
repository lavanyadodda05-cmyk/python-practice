```python

# INHERITANCE:like a family tree ,allows a new class (child)to inherit attributes and methods from an existing class(parent)
#  TYPES OF INHERITANCE
#    1.Single = one parent class one child class
#    2.multiple=multiple parents one child
#    3.multilevel = grand father to father to child
#    4.hirerical = multiple childs one parent
#    5.hybrid = combination of all

#   SINGLE INHERITANCE
class parent:
  def fun1(self):
    print("parent class")
class child(parent):
  def fun2(self):
    print("child class")
obj=child()
obj.fun1()
obj.fun2()

# MULTIPLE INHERITANCE
class father:
  def fun1(self):
    print("parent1 class")
class mother:
  def fun2(self):
    print("parent2 class")
class child(father,mother):
  def fun3(self):
    print("child class")
obj=child()
obj.fun1()
obj.fun2()
obj.fun3()

class Animal:
  def cat(self):
    print("eating")
class dog:
  def bark(self):
      print("barking")
class Hen(Animal,dog):
  def fly(self):
    print("flying")
d = Hen()
d.cat()
d.bark()
d.fly()

# MULTILEVEL INHERITANCE
class GrandFather:
  def fun1(self):
    print("grandfather class")
class Father(GrandFather):
  def fun2(self):
    print("father class")
class Son(Father):
  def fun3(self):
    print("son class")
obj=Son()
obj.fun1()
obj.fun2()
obj.fun3()

# HIERARICAL INHERITANCE
class parent:
  def fun1(self):
    print("parent1 class")
class child1(parent):
  def fun2(self):
    print("child1 class")
class child2(parent):
  def fun3(self):
    print("child2 class")
obj=child1()
obj.fun1()
obj.fun2()
obj1=child2()
obj1.fun3() 

# HYBRID INHERITANCE

class Animal:
    def eat(self):
        print("Eating")

class Mammal(Animal):  # Multilevel
    def walk(self):
        print("Walking")

class Pet:
    def play(self):
        print("Playing")

class Dog(Mammal, Pet):  # Multiple
    def bark(self):
        print("Woof!")

my_dog = Dog()
my_dog.eat()
my_dog.walk()
my_dog.play()
my_dog.bark()
```
