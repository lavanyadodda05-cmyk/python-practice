```python
# **C1. `price_with_tax`** *(Week 0)*

# **Functions & variables.** Write `price_with_tax(price, rate=0.1)` that returns the price plus tax, **rounded to 2 decimals**. `rate` has a default value of `0.1` (10%).
# Example: `price_with_tax(100)` → `110.0

def price_with_tax(price,rate=0.1):
    return round(price * (1+rate) , 2)

price_with_tax(100)

```

```python
# **C2. `bank_value`** *(Week 1)*

# **Conditionals.** Write `bank_value(greeting)` for a bank that gives a reward based on a greeting (ignore case and surrounding spaces):
# - starts with `"hello"` → `0`
# - otherwise, starts with `"h"` → `20`
# - anything else → `100`


def bank_value(greeting):
    greeting= greeting.strip().lower()
    if greeting.startswith("hello"):
        return 0
    elif greeting.startswith("h"):
        return 20
    else:
        return 100

bank_value("h")
```

```python
# **C4. `safe_divide`** *(Week 3)*

# **Exceptions.** Write `safe_divide(a, b)` that returns `a / b`, but:
# - returns `None` if `b` is zero (`ZeroDivisionError`)
# - returns the string `"invalid"` if the values can't be divided (`TypeError`)
# Use `try` / `except`.

def safe_divide(a,b):
    try:
        return a/b
    except ZeroDivisionError:
        return None
    except TypeError:
        return "invalid"
    
safe_divide(10,0)



```

```python
# **C5. `mean_length`** *(Week 4)*

# **Libraries.** Write `mean_length(words)` that uses the `statistics` module to return the **average length** of the words in a list.
# Example: `mean_length(["a", "bb", "ccc"])` → `2`

import statistics
def mean_length(words):
    return statistics.mean(len(word) for word in words)

mean_length(["ab","abb","abcd"])
```

```python
# **C9. `BankAccount`** *(Week 8)*

# **Object-oriented programming.** Create a class `BankAccount` with:
# - `__init__(self, owner, balance=0)` storing `owner` and `balance`
# - `deposit(amount)` – adds to the balance; raises `ValueError` if `amount <= 0`
# - `withdraw(amount)` – subtracts from the balance; raises `ValueError` if `amount` is more than the balance
# - `__str__` returning `"<owner>: $<balance>"`, e.g. `"Ada: $100"`
```

```python
class Bankaccount:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        if amount<=0:
            raise ValueError("amount must b ee positive")
        self.balance=self.balance+amount
    def withdraw(self,amount):
        if amount> self.balance:
            raise ValueError("insufficiant balance")
        self.balance=self.balance-amount

    def get_balance(self):
        return self.balance
acc=Bankaccount("lavanya",1000)
acc.deposit(800)
acc.withdraw(500)
print(acc.get_balance())
          


    


```
